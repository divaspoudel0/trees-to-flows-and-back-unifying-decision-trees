## Summary Table

```python
import json
from pathlib import Path

results_dir = Path('results')
summary = []

# Claim 1
d = json.loads((results_dir / 'claim1_theorems.json').read_text())
n_supported = sum(1 for v in d.values() if v['verdict'] == 'SUPPORTED')
summary.append(('Claim 1: Theorems', f'{n_supported}/{len(d)} theorems verified', 'PARTIAL' if n_supported < len(d) else 'SUPPORTED'))

# Claim 2
d = json.loads((results_dir / 'claim2_dsm_tree.json').read_text())
n_match = sum(1 for v in d.values() if v['matches_teacher_2pct'])
summary.append(('Claim 2: DSmTree', f'{n_match}/{len(d)} datasets match teacher', 'SUPPORTED' if n_match >= len(d) - 1 else 'PARTIAL'))

# Claim 3
d = json.loads((results_dir / 'claim3_implicit_tree.json').read_text())
valid = all(v['tree_discovery']['valid_dendrogram'] for v in d.values())
summary.append(('Claim 3: Implicit Trees', f'Dendrograms valid: {valid}', 'SUPPORTED' if valid else 'PARTIAL'))

# Claim 4
d = json.loads((results_dir / 'claim4_treeflow.json').read_text())
avg_wd = sum(v['wasserstein'] for v in d.values()) / len(d)
summary.append(('Claim 4: TreeFlow', f'Avg WD={avg_wd:.3f}', 'PARTIAL'))

# Claim 5
d = json.loads((results_dir / 'claim5_sample_complexity.json').read_text())
n_pass = sum(1 for v in d.values() if list(v.values())[-1])
summary.append(('Claim 5: Sample Complexity', f'{n_pass}/{len(d)} checks pass', 'SUPPORTED' if n_pass >= len(d) - 1 else 'PARTIAL'))

print(f'{"Claim":<35} {"Finding":<40} {"Verdict"}')
print('-' * 95)
for claim, finding, verdict in summary:
    print(f'{claim:<35} {finding:<40} {verdict}')
```
